```python
# Import pandas for data analysis
import pandas as pd

# Import numpy for numerical operations
import numpy as np

# Load the IMDB movie dataset
url ='https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv'
df = pd.read_csv(url)

# Display the first 5 rows
df.head()

```

```python
# Display number of rows and columns
print("Shape of dataset:", df.shape)
```

```python
# Display first 5 rows
df.head()
```

```python
# Display all column names
print(df.columns)
```

```python
# Display data type of each column
print(df.dtypes)
```

```python
# Count missing values in each column
print(df.isnull().sum())
```

```python
# Check the number of duplicate rows
print("Number of duplicate rows:", df.duplicated().sum())


# Remove duplicate rows
df = df.drop_duplicates()

# Check duplicates again
print("Duplicates after removal:", df.duplicated().sum())
```

```python
# Rename the Title column to movie title
df = df.rename(columns={"bahubali": "movie_title"})

# Display the updated column names
print(df.columns)
```

```python
# Get all genre values
genres = df["Genre"].dropna().str.split("|").explode().unique()

# Display unique genres
print(genres)
```

```python
# Count genres in each movie
df["Genre_Count"] = df["Genre"].fillna("").str.split("|").str.len()
df[["Title", "Genre", "Genre_Count"]].head()
```

```python
# Check missing values
print("Missing Revenue values:")
print(df["Revenue (Millions)"].isnull().sum())

print("\nMissing Metascore values:")
print(df["Metascore"].isnull().sum())
```

```python
# Fill missing Revenue values with the median revenue
df["Revenue (Millions)"] = df["Revenue (Millions)"].fillna(
    df["Revenue (Millions)"].median()
)

# Fill missing Metascore values with the median metascore
df["Metascore"] = df["Metascore"].fillna(
    df["Metascore"].median()
)

# Check missing values again
print(df[["Revenue (Millions)", "Metascore"]].isnull().sum())
```

```python
# Sort movies by Rating from highest to lowest
top_5_movies = df.sort_values(
    by="Rating",
    ascending=False
).head(5)

# Display movie title and rating
print(top_5_movies[["Title", "Rating"]])
```

```python
# Find the movie with the highest revenue
highest_revenue = df.loc[df["Revenue (Millions)"].idxmax()]

print("Movie with highest revenue:")
print(highest_revenue[["Title", "Revenue (Millions)"]])
```

```python
# Find the movie with the lowest revenue
lowest_revenue = df.loc[df["Revenue (Millions)"].idxmin()]

print("Movie with lowest revenue:")
print(lowest_revenue[["Title", "Revenue (Millions)"]])
```

```python
# Calculate average rating for each director
avg_rating_director = df.groupby("Director")["Rating"].mean()

# Sort from highest to lowest
avg_rating_director = avg_rating_director.sort_values(
    ascending=False
)

# Display the result
print(avg_rating_director)
```

```python
# Count the number of movies released in each year
movies_per_year = df.groupby("Year")["Title"].count()

# Display the result
print(movies_per_year)
```

```python
# Filter movies having:
# Rating greater than 8
# Revenue greater than 100 million

filtered_movies = df[
    (df["Rating"] > 8) &
    (df["Revenue (Millions)"] > 100)
]

# Display the movies
print(
    filtered_movies[
        ["Title", "Rating", "Revenue (Millions)"]
    ]
)
```

```python
# Group movies by Director
# Calculate average Rating and Revenue for each director

director_summary = df.groupby("Director").agg({
    "Rating": "mean",
    "Revenue (Millions)": "mean"
})

# Display the result
print(director_summary)
```

```python
# Group movies by Year
# Count number of movies
# Calculate average rating

year_summary = df.groupby("Year").agg(
    Movie_Count=("Title", "count"),
    Average_Rating=("Rating", "mean")
)

# Display the result
print(year_summary)
```

```python
#Create a new Profitability column
# Profitability = Revenue / Duration

df["Profitability"] = (
    df["Revenue (Millions)"] / df["Runtime (Minutes)"]
)

# Display the result
df[[
    "Title",
    "Revenue (Millions)",
    "Runtime (Minutes)",
    "Profitability"
]].head()
```

```python
# Sort movies by profitability
# and select the top 10

top_10_profitable = df.sort_values(
    by="Profitability",
    ascending=False
).head(10)

# Display the top 10
print(
    top_10_profitable[
        ["Title", "Profitability"]
    ]
)
```

```python
# Create a function to categorize movies based on rating

def rating_category(rating):

    # Rating below 6
    if rating < 6:
        return "Poor"

    # Rating from 6 to less than 7.5
    elif rating < 7.5:
        return "Average"

    # Rating 7.5 or above
    else:
        return "Excellent"
```

```python
# Create a custom function to calculate movie score

def movie_score(row):

    # Calculate score using Rating and Revenue
    score = (
        row["Rating"] * 10
        + row["Revenue (Millions)"]
    )

    return score
```

```python
# Apply the movie_score function to every row
df["Movie_Score"] = df.apply(
    movie_score,
    axis=1
)

# Display the result
df[[
    "Title",
    "Rating",
    "Revenue (Millions)",
    "Movie_Score"
]].head(10)
```
